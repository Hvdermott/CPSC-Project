```python
import numpy as np
from sklearn.preprocessing import StandardScaler
# ---------------------------
# log transformations
# ---------------------------
  """
  apply log-transformations for skewed distributions:
    - N2O emissions
    - TN
    - SOC
    - NO3-
    - NH4+

      constant (to avoid log(0)) ~ 1e-6

  """
def log_transform(arr, offset=1e-6):
    return np.log(arr + offset)
```

```python
# ---------------------------
# standardization
# ---------------------------
  """
 standardize array to mean=0, std=1

  """

def standardize(arr):
    scaler = StandardScaler()
    arr_flat = arr.reshape(-1, 1)
    arr_scaled = scaler.fit_transform(arr_flat)
    return arr_scaled.reshape(arr.shape), scaler
```

```python
# ---------------------------
# data compilation
# ---------------------------
  """
combine predictors into a single array for easier input
    * assumes transformations (log, standardization, normalization) are already applied.

  """

def compile_predictors(predictor_dict):

    compiled_data = {}
    for name, arr in predictor_dict.items():
        compiled_data[name] = arr.copy()
    return compiled_data
```

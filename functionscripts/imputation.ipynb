```python
# ---------------------------
# step 3: mean imputation
# ---------------------------
  """
 conduct mean imputation method to supplement missing values

  """

  miss_col= ["A","B","C","D","E","F"] # replace with missing cols
for col in miss_col:
    df[col].replace(np.nan,df[col].astype("float").mean(axis=0),inplace=True)

df["num-of-doors"].replace(np.nan,df["num-of-doors"].value_counts().idxmax(),inplace=True)
df.head().T
```
